```python
all_divs  = soup.find_all('div',class_='tuple ow-figma')
import numpy as np
import pandas as pd
```

```python
all_divs
```

```python
len(all_divs)
```

```python
Agency = []
origin  = []
origin_time = []
origin_date = []
duration = []
details = []
destination = []
reach_time = []
reach_date = []
price = []
flight_number = []

for flight in all_divs:
    try:
        Agency.append(flight.find('span',class_='i-b bold').text)
    except:
        Agency.append(np.nan)
    try:
        origin.append(flight.find('div',class_='depart-details').find('p',class_='fs-16 bold mob-origin').text)
    except:
        origin.append(np.nan)
    try:
        origin_time.append(flight.find('div',class_='depart-details').find('p',class_='fs-20 bold mob-time').text)
    except:
        origin_time.append(np.nan)
    try:
        origin_date.append(flight.find('div',class_='depart-details').find('p',class_='fs-12 font-grey bold mob-date').text)
    except:
        origin_date.append(np.nan)
    try:
        duration.append(flight.find('p',class_='fs-12 mob-duration').text)
    except:
        duration.append(np.nan)
    try:
        details.append(flight.find('span',class_='mob-duration').text)
    except:
        details.append(np.nan)
    try:
        destination.append(flight.find('div',class_='arrival-details text-right').find('p',class_='fs-16 bold mob-origin').text)
    except:
        destination.append(np.nan)
    try:
        reach_time.append(flight.find('div',class_='arrival-details text-right').find('p',class_='fs-20 bold mob-time').text)
    except:
        reach_time.append(np.nan)
    try:
        reach_date.append(flight.find('div',class_='arrival-details text-right').find('p',class_='fs-12 font-grey bold mob-date').text)
    except:
        reach_date.append(np.nan)
    try:
        price.append(flight.find('p',class_='ow-price-above-btn').text)
    except:
        price.append(np.nan)
    try:
        flight_number.append(flight.find('p',class_='normal fs-12 font-lightgrey fl-no').find('span').text)
    except:
        price.append(np.nan)
```

```python
len(flight_number)
```

```python
df = pd.DataFrame({
    'Company': Agency,
    'Flight Number' : flight_number,
    'From' : origin,
    'From Date':origin_date,
    'From Time':origin_time,
    'To': destination,
    'To Date':reach_date,
    'To Time':reach_time,
    'Details':details,
    'Duration' : duration,
    'Price':price
})
```

```python
df
```

```python
df['Price'] = df['Price'].str.replace(',','')
df['Price'] = df['Price'].str.replace('₹','').astype('int')
```

```python
df['Number of stop'] = df['Details'].apply(lambda x: x.strip().split()[0])
```

```python
df['Stop place'] = df['Details'].apply(lambda x: x.strip().split()[-1])
```

```python
df[df['Number of stop'] != '1']
```

```python
df.loc[df['Number of stop']=='Non','Number of stop'] = '0'
df.loc[df['Stop place'] =='Stop','Stop place'] = '-'
```

```python
df.info()
```

```python
df = df[['Company','Flight Number','From','From Date','From Time','To','To Date','To Time','Duration','Number of stop','Stop place','Price']]
```
